```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
from selenium.webdriver.common.action_chains import ActionChains

links = pd.read_excel("links_test.xlsx")

total = []  # 결과를 저장할 리스트

driver = webdriver.Chrome()

for url in links['links']:
    try:
        print(f"Accessing URL: {url}")
        driver.get(url)  # URL에 접속
        driver.implicitly_wait(10)
        time.sleep(2) 

        # 제목과 소개 정보 추출
        playnm = driver.find_element(By.TAG_NAME, "h3").text

        try:
            playintro = driver.find_element(By.CSS_SELECTOR, ".info1 .data").text
        except Exception as e:
            print(f"Intro not found for {url}: {e}")
            playintro = "N/A"  # 소개 정보가 없으면 기본값 설정

        total.append([playnm, playintro]) 
    
    except Exception as e:
        print(f"Error: {e}")

driver.quit()

# 리스트를 DataFrame으로 변환
df = pd.DataFrame(total, columns=["Title", "Introduction"])

# CSV 파일로 저장
output_path = "output.csv"
df.to_csv(output_path, index=False, encoding='utf-8-sig')  # UTF-8 인코딩 설정

print(f"Data saved to {output_path}")
```
