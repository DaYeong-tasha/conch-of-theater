```python
from bs4 import BeautifulSoup
from selenium.webdriver.support.ui import WebDriverWait
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
import pandas as pd
import GetCategory as cate
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.desired_capabilities import DesiredCapabilities
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
import time
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys


### Crawling Setup
# ChromeDriver 경로
chromedriver_path = "C:\\Users\\cokip\\Downloads\\chromedriver-win64\\chromedriver-win64\\chromedriver.exe" # 정확한 파일 경로 입력

# Service 객체 생성
service = Service(chromedriver_path)

# WebDriver 생성
browser = webdriver.Chrome(service=service)

# 암시적 대기
browser.implicitly_wait(3)

# 테스트 URL
# urls = cate.get_urls_list("http://mticket.interpark.com/Genre?Genre=DR")


total = []
for url in urls_list:
    browser = webdriver.Chrome()
    browser.implicitly_wait(3)
    browser.get(url)

    # 팝업 닫기 
    try:
        close_button = WebDriverWait(browser, 10).until(
            EC.presence_of_element_located((By.CLASS_NAME, "popupCloseBtn.is-bottomBtn"))
        )
        browser.execute_script("arguments[0].click();", close_button)
    except Exception as e:
        print(f"Error: {e}")

    while 1:
        try:
            # browser.find_element_by_class_name("moreBtn").send_keys(Keys.ENTER)
            browser.find_element(By.CLASS_NAME, "moreBtn").send_keys(Keys.ENTER)
            time.sleep(2)
        except:
            break



    try:
        browser.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
    
        playnm = browser.find_element(By.CLASS_NAME, "prdTitle").text
    
        gender_stats = []
        for k in browser.find_elements(By.CLASS_NAME, "statGenderType "):
            gender = k.find_element(By.CLASS_NAME, "statGenderValue").text.rstrip('%')
            gender_stats.append(gender)

        age_stats = []
        for x in browser.find_elements(By.CLASS_NAME, "statAgeType"):
            age = x.find_element(By.CLASS_NAME, "statAgePercent").text.rstrip('%')
            age_stats.append(age)

        row = [playnm] + gender_stats + age_stats
        total.append(row)
        print(f'{playnm}의 예매자 통계를 저장했습니다.')

        browser.close()
    except Exception as e:
                print(f"Error: {e}")
                
    except:
        print("no static")

print(total)
```
